```python
#genero nuevamente la variable pobre

# Tengo que generar un bucle que asigne valores separando primero por sexo y después por edad
# CH04 = 1 (hombre) o CH04 = 2 (mujer)
# CH06 (edad)

conds = [
    (EPH['CH04']==1) & (EPH['CH06']<1),
    (EPH['CH04']==1) & (EPH['CH06']==1),
    (EPH['CH04']==1) & (EPH['CH06']==2),
    (EPH['CH04']==1) & (EPH['CH06']==3),
    (EPH['CH04']==1) & (EPH['CH06']==4),
    (EPH['CH04']==1) & (EPH['CH06']==5),
    (EPH['CH04']==1) & (EPH['CH06']==6),
    (EPH['CH04']==1) & (EPH['CH06']==7),
    (EPH['CH04']==1) & (EPH['CH06']==8),
    (EPH['CH04']==1) & (EPH['CH06']==9),
    (EPH['CH04']==1) & (EPH['CH06']==10),
    (EPH['CH04']==1) & (EPH['CH06']==11),
    (EPH['CH04']==1) & (EPH['CH06']==12),
    (EPH['CH04']==1) & (EPH['CH06']==13),
    (EPH['CH04']==1) & (EPH['CH06']==14),
    (EPH['CH04']==1) & (EPH['CH06']==15),
    (EPH['CH04']==1) & (EPH['CH06']==16),
    (EPH['CH04']==1) & (EPH['CH06']==17),
    (EPH['CH04']==1) & (EPH['CH06'].between(18,29)),
    (EPH['CH04']==1) & (EPH['CH06'].between(30,45)),
    (EPH['CH04']==1) & (EPH['CH06'].between(46,60)),
    (EPH['CH04']==1) & (EPH['CH06'].between(61,75)),
    (EPH['CH04']==1) & (EPH['CH06']>75),
    (EPH['CH04']==2) & (EPH['CH06']<1),
    (EPH['CH04']==2) & (EPH['CH06']==1),
    (EPH['CH04']==2) & (EPH['CH06']==2),
    (EPH['CH04']==2) & (EPH['CH06']==3),
    (EPH['CH04']==2) & (EPH['CH06']==4),
    (EPH['CH04']==2) & (EPH['CH06']==5),
    (EPH['CH04']==2) & (EPH['CH06']==6),
    (EPH['CH04']==2) & (EPH['CH06']==7),
    (EPH['CH04']==2) & (EPH['CH06']==8),
    (EPH['CH04']==2) & (EPH['CH06']==9),
    (EPH['CH04']==2) & (EPH['CH06']==10),
    (EPH['CH04']==2) & (EPH['CH06']==11),
    (EPH['CH04']==2) & (EPH['CH06']==12),
    (EPH['CH04']==2) & (EPH['CH06']==13),
    (EPH['CH04']==2) & (EPH['CH06']==14),
    (EPH['CH04']==2) & (EPH['CH06']==15),
    (EPH['CH04']==2) & (EPH['CH06']==16),
    (EPH['CH04']==2) & (EPH['CH06']==17),
    (EPH['CH04']==2) & (EPH['CH06'].between(18,29)),
    (EPH['CH04']==2) & (EPH['CH06'].between(30,45)),
    (EPH['CH04']==2) & (EPH['CH06'].between(46,60)),
    (EPH['CH04']==2) & (EPH['CH06'].between(61,75)),
    (EPH['CH04']==2) & (EPH['CH06']>75)
]

choices = [0.35, 0.37, 0.46, 0.51, 0.55, 0.60, 0.64, 0.66, 0.68, 0.69,
          0.79, 0.82, 0.85, 0.90, 0.96, 1.00, 1.03, 1.04, 
           1.02, 1.00, 1.00, 0.83, 0.74, #Hasta acá hombres
          0.35, 0.37, 0.46, 0.51, 0.55, 0.60, 0.64, 0.66, 0.68, 0.69,
          0.70, 0.72, 0.74, 0.76, 0.76, 0.77, 0.77, 0.77,
          0.76, 0.77, 0.76, 0.67, 0.63]

EPH['adulto_equiv'] = np.select(conds, choices, default=np.nan)


# Agrupar por CODUSU y NRO_HOGAR, y sumar los valores de adulto_equiv
EPH['ad_equiv_hogar'] = EPH.groupby(['CODUSU', 'NRO_HOGAR'])['adulto_equiv'].transform('sum')


cbt_2005 = 205.07
cbt_2025 = 365177

# Ingreso necesario
EPH["ingreso_necesario"] = np.where(
    EPH["ANIO"] == 2005,
    EPH["ad_equiv_hogar"] * cbt_2005,
    EPH["ad_equiv_hogar"] * cbt_2025
)

# Definir pobreza
EPH["pobre"] = np.where(
    (EPH["ITF"] != 0) & (EPH["ITF"] < EPH["ingreso_necesario"]),
    1, 0
)
```

## Ejercicio 2

```python
# Ya que en el tp anterior no se consideraron las variables, agrego CH12, CH13 y CH14 a mi base

# Cargo las bases orginales
eph2005 = pd.read_stata("Individual_t105.dta")
eph2025 = pd.read_excel("usu_individual_T125.xlsx")
```

```python
eph2005.columns = eph2005.columns.str.upper()
```

```python
# Filtrar solo región Noroeste
eph2005 = eph2005[eph2005["REGION"]=="NOA"].copy()
eph2025 = eph2025[eph2025["REGION"]== 40].copy()
```

```python
#Me aseguro que el año sea entero
EPH["ANIO"] = EPH["ANIO"].astype("int64")
```

```python
#Genero una lista de las variables que necesito

vars_educ = ["CODUSU", "ANO4", "CH12", "CH13", "CH14"]

educ2005 = eph2005[vars_educ].rename(columns={"ANO4":"ANIO"}).copy()
educ2025 = eph2025[vars_educ].rename(columns={"ANO4":"ANIO"}).copy()


# Recodificación 2005 (de palabras a números)

map_ch12 = {
    "Jardín/Preescolar": 1,
    "Primario": 2,
    "EGB": 3,
    "Secundario": 4,
    "Polimodal": 5,
    "Terciario": 6,
    "Universitario": 7,
    "Posgrado Universitario": 8,
    "Educación especial (discapacitado)": 9
}

map_ch13 = {
    "Sí": 1,
    "No": 2,
    "Ns./Nr.": 9,
}

educ2005["CH12"] = educ2005["CH12"].map(map_ch12)
educ2005["CH13"] = educ2005["CH13"].map(map_ch13)

# CH14: quitar ceros a la izquierda 
educ2005["CH14"] = educ2005["CH14"].astype(str).str.lstrip("0").replace({
    "": "0"  # Para casos donde era '00'
})

# Para educ2025, solo quitar ceros a la izquierda
educ2025["CH14"] = educ2025["CH14"].astype(str).str.lstrip("0")


#  Unificar tipo de ANIO

educ2005["ANIO"] = educ2005["ANIO"].astype(int)
educ2025["ANIO"] = educ2025["ANIO"].astype(int)


#  Eliminar duplicados por persona por año

educ2005 = educ2005.groupby(['CODUSU','ANIO'], as_index=False).first()
educ2025 = educ2025.groupby(['CODUSU','ANIO'], as_index=False).first()


#  Concatenar 2005 y 2025

educacion = pd.concat([educ2005, educ2025], ignore_index=True, sort=False)


#  Merge con la base combinada EPH

merge_keys = ["CODUSU","ANIO"]
EPH_merged = EPH.merge(educacion, on=merge_keys, how='left')


# 7. Chequeo final de valores

for col in ["CH12","CH13","CH14"]:
    print(f"\n{col}:")
    print(EPH_merged[col].value_counts(dropna=False))
```

```python
EPH_merged
```

```python
def calcular_educ(row):
    nivel = row["CH12"]
    completo = row["CH13"]
    try:
        ultimo = int(row["CH14"])
    except:
        ultimo = np.nan

    if nivel == 9:  # Educación especial
        return np.nan  

    if nivel == 1:  # Jardín/preescolar
        return 0  

    if nivel == 2:  # Primario
        return 6 if completo == 1 else (ultimo if 1 <= ultimo <= 6 else np.nan)

    if nivel == 3:  # EGB
        return 9 if completo == 1 else (ultimo if 1 <= ultimo <= 9 else np.nan)

    if nivel in [4,5]:  # Secundario/Polimodal
        return 12 if completo == 1 else (6 + ultimo if 1 <= ultimo <= 6 else np.nan)

    if nivel == 6:  # Terciario
        return 15 if completo == 1 else (12 + ultimo if 1 <= ultimo <= 3 else np.nan)

    if nivel == 7:  # Universitario
        return 17 if completo == 1 else (12 + ultimo if 1 <= ultimo <= 5 else np.nan)

    if nivel == 8:  # Posgrado
        return 19 if completo == 1 else (17 + ultimo if 1 <= ultimo <= 2 else np.nan)

    return np.nan

# Aplicar la función
EPH["educ"] = EPH_merged.apply(calcular_educ, axis=1)
```

```python
desc_educ = EPH["educ"].describe(percentiles=[0.5]).to_frame().T
desc_educ = desc_educ.rename(index={"educ": "Años de educación"})
print(desc_educ[["count","mean","std","min","50%","max"]])
```
